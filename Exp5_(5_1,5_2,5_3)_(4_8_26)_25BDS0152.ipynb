```python
import pandas as pd
df = pd.read_csv('/content/fraudTrain.csv')
```

```python
df.info()
df['is_fraud'].value_counts()
```

```python
df.head(10)
```

```python
Q1 = df['amt'].quantile(0.25)
Q3 = df['amt'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df['outlier_iqr'] = ((df['amt'] < lower) | (df['amt'] > upper)).astype(int)

print("Total outliers found:", df['outlier_iqr'].sum())
```

```python
pd.crosstab(df['outlier_iqr'], df['is_fraud'])
```

```python
import matplotlib.pyplot as plt

plt.boxplot(df['amt'])
plt.title("Boxplot of transaction amount")
plt.show()
```
