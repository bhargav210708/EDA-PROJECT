```python
import numpy as np
import pandas as pd
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer,SimpleImputer
df = pd.DataFrame({

                   "Character ":['Tony','Peter','Thor',np.nan,np.nan],
                  "Age" :[np.nan,22,1500,47,np.nan],
                   "Experience":[10,5,1300,np.nan,np.nan],


})
```

```python
mean_imputer = SimpleImputer(strategy="mean")
df["Age_Mean"] = mean_imputer.fit_transform(df[["Age"]])
print(df["Age_Mean"])
```

```python
median_imputer = SimpleImputer(strategy="median")
df["Experience_Median"] = median_imputer.fit_transform(df[["Experience"]])
print(df["Experience_Median"])
```

```python
mode_imputer = SimpleImputer(strategy="most_frequent")
df["Age_Mode"] = mode_imputer.fit_transform(df[["Age"]])
print(df["Age_Mode"])
```

```python
num_cols = ["Age", "Experience"]
mice_imputer = IterativeImputer(max_iter=10, random_state=42)
df_mice = df[num_cols].copy()
df_mice.iloc[:, :] = mice_imputer.fit_transform(df_mice)
df["Age_MICE"] = df_mice["Age"]
df["Expeience_MICE"] = df_mice["Experience"]
print(df)
```
